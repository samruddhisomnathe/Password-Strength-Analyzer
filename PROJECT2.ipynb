```python
import re

def analyze_password(password):
    score = 0

    # Length Check
    if len(password) >= 8:
        score += 1

    # Uppercase Check
    if re.search(r"[A-Z]", password):
        score += 1

    # Lowercase Check
    if re.search(r"[a-z]", password):
        score += 1

    # Digit Check
    if re.search(r"\d", password):
        score += 1

    # Special Character Check
    if re.search(r"[!@#$%^&*(),.?\":{}|<>]", password):
        score += 1

    # Strength Evaluation
    if score <= 2:
        strength = "Weak"
    elif score == 3 or score == 4:
        strength = "Medium"
    else:
        strength = "Strong"

    print("\nPassword Analysis")
    print("------------------")
    print("Password Strength:", strength)
    print("Score:", score, "/5")

    if strength == "Weak":
        print("Suggestion: Use uppercase, lowercase, numbers, and special characters.")
    elif strength == "Medium":
        print("Suggestion: Add more complexity to make it stronger.")
    else:
        print("Excellent! Your password is strong.")

# Main Program
password = input("Enter Password: ")
analyze_password(password)
```
